## Introduction

### GenAI Terminology

- **Large Language Models (LLMs)**: Neural networks to
understand and generate text and other forms of data.
- **Generative Pre-trained Transformer (GPT)**: A specific type of LLM architecture that uses the transformer neural network, is pre-trained on large text datasets, and generates text by predicting the next token in sequence.
- **Training**: Teaching a model from scratch how to perform
a task using large amounts of data.
- **Retraining**: Retraining a base model with a new dataset.
- **Finetuning**: Teaching a base model specialized knowledge
by adjusting internal parameters such as weights and
biases.
- **Retrieval Augmented Generation (RAG)**: Utilizing
external data to augment an LLM’s knowledge without
changing model parameters.
- **Pipeline**: A series of connected stages that transform data, interact with an LLM, and produce a desired output
- **Tailoring**: Adapting or customizing a general-purpose LLM to perform specific tasks or generate specialized outputs for a particular domain or application.
- **Reinforcement Learning from Human Feedback (RLHF)**:
An iterative technique used to train AI models by
incorporating human feedback to improve their
performance and alignment with human preferences.

## Interactive Systems Engineering Expert with Token Metrics

````python
import sys
import time
from google import genai

# 1. Verification of Client Credentials
if not 'api_key' in globals() or not api_key:
    raise ValueError("Please verify your 'api_key' is defined in the notebook.")

client = genai.Client(api_key=api_key)
MODEL_NAME = gemini_model if 'gemini_model' in globals() else 'gemini-2.5-flash'

# 2. Establish persistent, state-aware chat session
chat = client.chats.create(
    model=MODEL_NAME,
    config={
        "system_instruction": "You are a systems engineering expert assistant.",
        "temperature": 0.3
    }
)

print(f"""
============================================================
Google GenAI Client Active | Model: {MODEL_NAME}
Type 'exit', 'quit', or press Ctrl+C to end the session.
============================================================
""")

# 3. Terminal Interface Loop
try:
    while True:
        user_input = input("\nYou  ")

        if user_input.strip().lower() in ["exit", "quit"]:
            print("\nShutting down interactive session. Goodbye!")
            break

        if not user_input.strip():
            continue

        print("\nThinking...", end="", flush=True)

        try:
            start_time = time.perf_counter()
            # Sends the user turn and retrieves the state-aware response
            response = chat.send_message(user_input)
            elapsed_time = time.perf_counter() - start_time

            # Extract response payload and token usage metadata
            ai_reply = response.text
            usage = response.usage_metadata
            prompt_tokens = usage.prompt_token_count if usage else 0
            completion_tokens = usage.candidates_token_count if usage else 0
            total_tokens = usage.total_token_count if usage else 0
            tokens_per_sec = completion_tokens / elapsed_time if elapsed_time > 0 else 0.0

            # Print results safely
            print(f"\rGemini  {ai_reply}\n")
            print(
                f"[Metrics] Input: {prompt_tokens} tok | "
                f"Output: {completion_tokens} tok | "
                f"Total: {total_tokens} tok | "
                f"Speed: {tokens_per_sec:.2f} tok/s (Time: {elapsed_time:.2f}s)"
            )
        except Exception as api_err:
            print(f"\r[!] API Call Error: {api_err}")

except KeyboardInterrupt:
    print("\n\nSession interrupted via hotkey. Goodbye!")
````

#### Constrain LLM Output for Control System Component Classification

````python
# Constrain LLM Output for Control System Component Classification
from google import genai
from google.colab import userdata
import enum

# Define a comprehensive enumeration for Industrial Control System (ICS) Component Types
class IcsComponentType(enum.Enum):
    VALVE = "Valve"
    SENSOR_TEMPERATURE = "Temperature Sensor"
    SENSOR_PRESSURE = "Pressure Sensor"
    ACTUATOR_MOTOR = "Motor Actuator"
    ACTUATOR_SOLENOID = "Solenoid Actuator"
    CONTROLLER_PLC = "PLC Controller"
    CONTROLLER_DCS = "DCS Controller"
    COMM_GATEWAY = "Communication Gateway"
    POWER_SUPPLY = "Power Supply Unit"
    HMI_PANEL = "Human-Machine Interface Panel"
    NETWORK_SWITCH = "Network Switch"
    CIRCUIT_BREAKER = "Circuit Breaker"
    PUMP = "Pump"
    FLOW_METER = "Flow Meter"
    TRANSDUCER = "Transducer"

client = genai.Client(api_key=userdata.get('api_key'))

# Example query: A natural language description of a faulty component
component_description = "electricity come from it"

response = client.models.generate_content(
    model=gemini_model, # Or a more capable model if needed for complex reasoning
    contents=f'Classify the following industrial component based on its function: "{component_description}"',
    config={
        'response_mime_type': 'text/x.enum',
        'response_schema': IcsComponentType, # Constrain output to our defined enum
    },
)

# Print the classified component type
print(f"Original description: \"{component_description}\"")
print(f"Classified Component Type: {response.text}")

# Example of how a downstream system might use this:
if response.text == IcsComponentType.SENSOR_TEMPERATURE.value:
    print("Action: Initiate temperature sensor diagnostics protocol and check spare parts inventory for 'THERM-001'.")
elif response.text == IcsComponentType.VALVE.value:
    print("Action: Check valve actuator status and prepare for mechanical inspection.")
else:
    print("Action: Refer to general component troubleshooting guide.")
````

````python
import sys
import time
from google import genai

# 1. Verification of Client Credentials
if not 'api_key' in globals() or not api_key:
    raise ValueError("Please verify your 'api_key' is defined in the notebook.")

client = genai.Client(api_key=api_key)
MODEL_NAME = gemini_model if 'gemini_model' in globals() else 'gemini-2.5-flash'

# 2. Establish persistent, state-aware chat session
chat = client.chats.create(
    model=MODEL_NAME,
    config={
        "system_instruction": "You are a systems engineering expert assistant.",
        "temperature": 0.3
    }
)

print(f"""
============================================================
Google GenAI Client Active | Model: {MODEL_NAME}
Type 'exit', 'quit', or press Ctrl+C to end the session.
============================================================
""")

# 3. Terminal Interface Loop
try:
    while True:
        user_input = input("\nYou 👉 ")

        if user_input.strip().lower() in ["exit", "quit"]:
            print("\nShutting down interactive session. Goodbye!")
            break

        if not user_input.strip():
            continue

        print("\nThinking...", end="", flush=True)

        try:
            start_time = time.perf_counter()
            # Sends the user turn and retrieves the state-aware response
            response = chat.send_message(user_input)
            elapsed_time = time.perf_counter() - start_time

            # Extract response payload and token usage metadata
            ai_reply = response.text
            usage = response.usage_metadata
            prompt_tokens = usage.prompt_token_count if usage else 0
            completion_tokens = usage.candidates_token_count if usage else 0
            total_tokens = usage.total_token_count if usage else 0
            tokens_per_sec = completion_tokens / elapsed_time if elapsed_time > 0 else 0.0

            # Print results safely
            print(f"\rGemini 👉 {ai_reply}\n")
            print(
                f"[Metrics] Input: {prompt_tokens} tok | "
                f"Output: {completion_tokens} tok | "
                f"Total: {total_tokens} tok | "
                f"Speed: {tokens_per_sec:.2f} tok/s (Time: {elapsed_time:.2f}s)"
            )
        except Exception as api_err:
            print(f"\r[!] API Call Error: {api_err}")

except KeyboardInterrupt:
    print("\n\nSession interrupted via hotkey. Goodbye!")
````

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Universal OpenAI SDK Chat Shell
Works with local Ollama, LM Studio, or official cloud OpenAI models.

import sys
import time
from openai import OpenAI

# 1. Configuration
# --- Local Ollama Endpoint ---
BASE_URL = "http://localhost:11434/v1"
API_KEY = "ollama"  # Standard placeholder required by the SDK
MODEL_NAME = "llama3.3:70b"

# API_KEY = "sk-proj-..."
# MODEL_NAME = "gpt-4o"

# Initialize client
client = OpenAI(base_url=BASE_URL, api_key=API_KEY) if BASE_URL else OpenAI(api_key=API_KEY)

# 2. Establish session history
history = [
    {
        "role": "system",
        "content": "You are a systems engineering expert."
    }
]

print(f"""
============================================================
OpenAI Client Active | Model: {MODEL_NAME} | Host: {BASE_URL or 'OpenAI Cloud'}
Type 'exit', 'quit', or press Ctrl+C to end the session.
============================================================
""")

# 3. Continuous Shell Loop
try:
    while True:
        user_input = input("\nYou 👉 ")
        
        if user_input.strip().lower() in ["exit", "quit"]:
            print("\nShutting down session. Goodbye!")
            break
            
        if not user_input.strip():
            continue
            
        history.append({"role": "user", "content": user_input})
        print("\n🛠️AI Thinking...", end="", flush=True)
        
        try:
            start_time = time.perf_counter()
            
            # OpenAI-compatible API call
            response = client.chat.completions.create(
                model=MODEL_NAME,
                messages=history,
                temperature=0.3
            )
            
            elapsed_time = time.perf_counter() - start_time
            
            # Extract content and token metadata
            ai_reply = response.choices[0].message.content
            usage = response.usage
            prompt_tokens = usage.prompt_tokens if usage else 0
            completion_tokens = usage.completion_tokens if usage else 0
            total_tokens = usage.total_tokens if usage else 0
            
            tokens_per_sec = completion_tokens / elapsed_time if elapsed_time > 0 else 0.0

            # Clear line and render output
            print(f"\r🛠️AI {ai_reply}\n")
            
            print(
                f"📊 [Metrics] Input Tokens: {prompt_tokens} | "
                f"Output Tokens: {completion_tokens} | "
                f"Total Tokens: {total_tokens} | "
                f"Time: {elapsed_time:.2f}s ({tokens_per_sec:.2f} tok/s)"
            )
            
            history.append({"role": "assistant", "content": ai_reply})
            
        except Exception as api_err:
            print(f"\r[!] API Communication Error: {api_err}")
            history.pop()

except KeyboardInterrupt:
    print("\n\nSession interrupted via terminal shortcut. Goodbye!")
    sys.exit(0)

```

# Integrating LLMs with External Tools Through APIs

- Systems engineering practice increasingly benefits from AI assistants that can interact with external tools rather than rely only on internal language generation
- API-based integration allows an AI assistant to retrieve current data, invoke analytic services, and return results grounded in deterministic computation
- This approach is especially valuable when the task requires traceability, reproducibility, and quantitative rigor
- It supports a disciplined separation between conversational interaction and formal engineering computation
- The basic principle is to assign interpretation and synthesis to the AI while assigning authoritative calculation or retrieval to the external tool

## YAML
YAML is a human-readable data format used to define APIs.  It is like JSON but easier to read, and is often used to define configurations, settings, or structured data. These include application or tool configurations, AI model behavior and metadata, or input parameters for a system. They are organized with key-value pairs and indentation to represent hierarchy.

### GPT Caller Instructions

```
....

Use the named schema operation `ReliabilityCalculate` to invoke the external
tool. Construct the request with exactly these three parameters:

{
  "mtbf": <float>,
  "mission_time": <float>,
  "redundancy": <integer>
}

Where:
- `mtbf` is Mean Time Between Failures in hours
- `mission_time` is mission duration in hours
- `redundancy` is the number of identical components operating in parallel and must be an integer >= 1

Do not compute the reliability internally. Always call the external tool once
these parameters are known or reasonably estimated.

After the tool returns, present:
- component reliability
- system reliability
- system failure probability

Also include:
- the exact calculation source URL returned by the tool
- the returned fault-tree graphic rendered in a fenced `svg` block
- any assumptions you made to derive `mtbf`, `mission_time`, or `redundancy`
```

### Reliability Modeling YAML Example
 This is the named `ReliabilityCalculate` schema referred to in the instructions. It formally specifies the interface contract between the GPT and the external domain tool.

```yaml
openapi: 3.1.0
servers:
  - url: https://se-lib.org/online/
paths:
  /reliability.py:
    get:
      operationId: ReliabilityCalculate
      parameters:
        - in: query
          name: mtbf
          description: Mean Time Between Failures in hours.
          required: true
          schema:
            type: number
            minimum: 0
        - in: query
          name: mission_time
          description: Mission time in hours
          required: true
          schema:
            type: number
            minimum: 0
        - in: query
          name: redundancy
          description: the number of identical components operating in parallel (integer ≥ 1)
          required: true
          schema:
            type: number
            minimum: 1

```
